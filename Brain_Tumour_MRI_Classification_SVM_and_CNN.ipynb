```
!pip install kagglehub -q

import kagglehub
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import seaborn as sns
import os
import shutil
import random
import warnings
warnings.filterwarnings('ignore')

import tensorflow as tf
from tensorflow.keras import layers, models, callbacks
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.preprocessing.image import (ImageDataGenerator,
                                                   load_img, img_to_array)
from tensorflow.keras.utils import to_categorical

from sklearn.svm import SVC
from sklearn.decomposition import PCA
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import (classification_report, confusion_matrix,
                             f1_score, roc_curve, auc)

print("TensorFlow version:", tf.__version__)
print("GPU Available:", tf.config.list_physical_devices('GPU'))
print("All libraries loaded successfully")


path = kagglehub.dataset_download("masoudnickparvar/brain-tumor-mri-dataset")
print("Dataset downloaded to:", path)

full_train_dir = os.path.join(path, "Training")
full_test_dir  = os.path.join(path, "Testing")
classes        = sorted(os.listdir(full_train_dir))
print("Classes:", classes)




random.seed(42)
TRAIN_SAMPLES = 500
TEST_SAMPLES  = 100

def create_subset(src_dir, dst_dir, samples_per_class):
    if os.path.exists(dst_dir):
        shutil.rmtree(dst_dir)
    os.makedirs(dst_dir)
    for cls in sorted(os.listdir(src_dir)):
        src_cls = os.path.join(src_dir, cls)
        dst_cls = os.path.join(dst_dir, cls)
        os.makedirs(dst_cls)
        imgs     = os.listdir(src_cls)
        selected = random.sample(imgs, min(samples_per_class, len(imgs)))
        for img in selected:
            shutil.copy(os.path.join(src_cls, img),
                        os.path.join(dst_cls, img))
    print(f"Subset created: {dst_dir}")


train_dir = '/content/train_subset'
val_dir   = '/content/val_subset'
test_dir  = '/content/test_subset'


def create_train_val_split(src_dir, train_dst, val_dst,
                           train_per_class=400, val_per_class=100):
    for dst in [train_dst, val_dst]:
        if os.path.exists(dst):
            shutil.rmtree(dst)
        os.makedirs(dst)

    for cls in sorted(os.listdir(src_dir)):
        src_cls   = os.path.join(src_dir, cls)
        imgs      = random.sample(os.listdir(src_cls),
                                  min(train_per_class + val_per_class,
                                      len(os.listdir(src_cls))))
        train_imgs = imgs[:train_per_class]
        val_imgs   = imgs[train_per_class:]

        os.makedirs(os.path.join(train_dst, cls))
        os.makedirs(os.path.join(val_dst,   cls))

        for img in train_imgs:
            shutil.copy(os.path.join(src_cls, img),
                        os.path.join(train_dst, cls, img))
        for img in val_imgs:
            shutil.copy(os.path.join(src_cls, img),
                        os.path.join(val_dst, cls, img))

    print(f"Train/Val split done → {train_dst} | {val_dst}")

create_train_val_split(full_train_dir, train_dir, val_dir)
create_subset(full_test_dir, test_dir, TEST_SAMPLES)


for cls in classes:
    tr = len(os.listdir(os.path.join(train_dir, cls)))
    vl = len(os.listdir(os.path.join(val_dir,   cls)))
    te = len(os.listdir(os.path.join(test_dir,  cls)))
    print(f"  {cls}: {tr} train | {vl} val | {te} test")




train_counts = {cls: len(os.listdir(os.path.join(train_dir, cls))) for cls in classes}
test_counts  = {cls: len(os.listdir(os.path.join(test_dir,  cls))) for cls in classes}

print("\nTraining samples:", train_counts)
print("Testing samples:",  test_counts)


fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].bar(train_counts.keys(), train_counts.values(), color='steelblue')
axes[0].set_title('Training Set Class Distribution')
axes[0].set_xlabel('Tumour Class')
axes[0].set_ylabel('Number of Images')

axes[1].bar(test_counts.keys(), test_counts.values(), color='coral')
axes[1].set_title('Testing Set Class Distribution')
axes[1].set_xlabel('Tumour Class')
axes[1].set_ylabel('Number of Images')
plt.tight_layout()
plt.savefig('class_distribution.png', dpi=150)
plt.show()


fig, axes = plt.subplots(1, 4, figsize=(16, 4))
for i, cls in enumerate(classes):
    img_path = os.path.join(train_dir, cls,
                 os.listdir(os.path.join(train_dir, cls))[0])
    img = load_img(img_path, target_size=(128, 128))
    axes[i].imshow(img)
    axes[i].set_title(cls)
    axes[i].axis('off')
plt.suptitle('Sample MRI Images per Class', fontsize=14)
plt.tight_layout()
plt.savefig('sample_images.png', dpi=150)
plt.show()


plt.figure(figsize=(12, 5))
for cls in classes:
    pixel_vals = []
    cls_path   = os.path.join(train_dir, cls)
    for img_name in os.listdir(cls_path)[:50]:
        img = load_img(os.path.join(cls_path, img_name),
                       target_size=(128, 128), color_mode='grayscale')
        pixel_vals.extend(img_to_array(img).flatten() / 255.0)
    sns.kdeplot(pixel_vals, label=cls)
plt.title('Pixel Intensity Distribution per Class')
plt.xlabel('Normalised Pixel Value')
plt.ylabel('Density')
plt.legend()
plt.tight_layout()
plt.savefig('pixel_intensity.png', dpi=150)
plt.show()


datagen_preview = ImageDataGenerator(
    rotation_range=20, zoom_range=0.2,
    width_shift_range=0.2, height_shift_range=0.2,
    horizontal_flip=True
)
sample_path = os.path.join(train_dir, classes[0],
               os.listdir(os.path.join(train_dir, classes[0]))[0])
sample_arr  = img_to_array(load_img(sample_path, target_size=(128, 128)))
sample_exp  = sample_arr.reshape((1,) + sample_arr.shape)

fig, axes = plt.subplots(1, 5, figsize=(16, 4))
for i, batch in enumerate(datagen_preview.flow(sample_exp, batch_size=1)):
    axes[i].imshow(batch[0].astype('uint8'))
    axes[i].axis('off')
    axes[i].set_title(f'Aug {i+1}')
    if i == 4:
        break
plt.suptitle('Data Augmentation Examples', fontsize=14)
plt.tight_layout()
plt.savefig('augmentation.png', dpi=150)
plt.show()

print("EDA Complete")




IMG_SIZE    = (128, 128)
BATCH_SIZE  = 32
NUM_CLASSES = 4


train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20, zoom_range=0.2,
    width_shift_range=0.2, height_shift_range=0.2,
    horizontal_flip=True
)
val_test_datagen = ImageDataGenerator(rescale=1./255)

train_gen = train_datagen.flow_from_directory(
    train_dir, target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=True
)
val_gen = val_test_datagen.flow_from_directory(
    val_dir, target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=False
)
test_gen = val_test_datagen.flow_from_directory(
    test_dir, target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=False
)

class_names = list(train_gen.class_indices.keys())
print("Class indices:", train_gen.class_indices)


early_stop = callbacks.EarlyStopping(
    monitor='val_loss', patience=5,
    restore_best_weights=True, verbose=1
)
reduce_lr = callbacks.ReduceLROnPlateau(
    monitor='val_loss', factor=0.5, patience=3, verbose=1
)


inputs = tf.keras.Input(shape=(128, 128, 3))

x = layers.Conv2D(32,  (3,3), activation='relu', name='conv1')(inputs)
x = layers.BatchNormalization()(x)
x = layers.MaxPooling2D(2, 2)(x)

x = layers.Conv2D(64,  (3,3), activation='relu', name='conv2')(x)
x = layers.BatchNormalization()(x)
x = layers.MaxPooling2D(2, 2)(x)

x = layers.Conv2D(128, (3,3), activation='relu', name='conv3')(x)
x = layers.BatchNormalization()(x)
x = layers.MaxPooling2D(2, 2)(x)

x = layers.Conv2D(256, (3,3), activation='relu', name='conv4_last')(x)
x = layers.BatchNormalization()(x)
x = layers.MaxPooling2D(2, 2)(x)

x = layers.GlobalAveragePooling2D()(x)
x = layers.Dense(256, activation='relu')(x)
x = layers.Dropout(0.5)(x)
outputs = layers.Dense(NUM_CLASSES, activation='softmax')(x)

cnn_model = tf.keras.Model(inputs, outputs, name='custom_cnn')
cnn_model.compile(
    optimizer=Adam(learning_rate=1e-3),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
cnn_model.summary()

print("\nTraining Custom CNN...")
cnn_history = cnn_model.fit(
    train_gen, validation_data=val_gen,
    epochs=20, callbacks=[early_stop, reduce_lr]
)
cnn_model.save('cnn_model.keras')
print("CNN saved")


print("\nPreparing SVM data...")

def load_images_for_svm(directory, img_size=64):
    images, labels = [], []
    for cls in sorted(os.listdir(directory)):
        cls_path = os.path.join(directory, cls)
        for img_name in os.listdir(cls_path):
            img = load_img(os.path.join(cls_path, img_name),
                           target_size=(img_size, img_size),
                           color_mode='grayscale')
            images.append(img_to_array(img).flatten() / 255.0)
            labels.append(cls)
    return np.array(images), np.array(labels)

X_train_svm, y_train_svm = load_images_for_svm(train_dir)
X_test_svm,  y_test_svm  = load_images_for_svm(test_dir)

pca         = PCA(n_components=100)
X_train_pca = pca.fit_transform(X_train_svm)
X_test_pca  = pca.transform(X_test_svm)

le          = LabelEncoder()
y_train_enc = le.fit_transform(y_train_svm)
y_test_enc  = le.transform(y_test_svm)

print("Running SVM Grid Search...")
param_grid = {'C': [1, 10], 'gamma': ['scale']}
svm = GridSearchCV(SVC(kernel='rbf', probability=True),
                   param_grid, cv=3, verbose=1, n_jobs=-1)
svm.fit(X_train_pca, y_train_enc)
print("Best SVM params:", svm.best_params_)

print("Methodology Complete")



cnn_loss, cnn_acc = cnn_model.evaluate(test_gen, verbose=0)
print(f"\nCustom CNN  — Accuracy: {cnn_acc:.4f} | Loss: {cnn_loss:.4f}")

svm_preds = svm.predict(X_test_pca)
svm_acc   = np.mean(svm_preds == y_test_enc)
print(f"SVM         — Accuracy: {svm_acc:.4f}")

test_gen.reset()
cnn_preds = np.argmax(cnn_model.predict(test_gen), axis=1)


def plot_confusion_matrix(y_true, y_pred, title, labels):
    cm_vals = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm_vals, annot=True, fmt='d', cmap='Blues',
                xticklabels=labels, yticklabels=labels)
    plt.title(f'Confusion Matrix — {title}')
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.tight_layout()
    plt.savefig(f'cm_{title.replace(" ","_")}.png', dpi=150)
    plt.show()

plot_confusion_matrix(test_gen.classes, cnn_preds,    'Custom CNN', class_names)
plot_confusion_matrix(y_test_enc,       svm_preds,    'SVM',        le.classes_)


print("\n--- Custom CNN ---")
print(classification_report(test_gen.classes, cnn_preds, target_names=class_names))

print("\n--- SVM ---")
print(classification_report(y_test_enc, svm_preds, target_names=le.classes_))


def plot_learning_curves(history, title):
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    axes[0].plot(history.history['accuracy'],     label='Train')
    axes[0].plot(history.history['val_accuracy'], label='Val')
    axes[0].set_title(f'{title} — Accuracy')
    axes[0].set_xlabel('Epoch')
    axes[0].set_ylabel('Accuracy')
    axes[0].legend()

    axes[1].plot(history.history['loss'],     label='Train')
    axes[1].plot(history.history['val_loss'], label='Val')
    axes[1].set_title(f'{title} — Loss')
    axes[1].set_xlabel('Epoch')
    axes[1].set_ylabel('Loss')
    axes[1].legend()

    plt.tight_layout()
    plt.savefig(f'lc_{title.replace(" ","_")}.png', dpi=150)
    plt.show()

plot_learning_curves(cnn_history,       'Custom CNN')


models_list = ['SVM', 'Custom CNN']
accuracies  = [svm_acc, cnn_acc]
colours     = ['coral', 'steelblue']

plt.figure(figsize=(8, 5))
bars = plt.bar(models_list, accuracies, color=colours)
plt.ylim(0.4, 1.05)
plt.title('Model Accuracy Comparison')
plt.ylabel('Test Accuracy')
for bar, acc in zip(bars, accuracies):
    plt.text(bar.get_x() + bar.get_width()/2,
             bar.get_height() + 0.01,
             f'{acc:.2%}', ha='center', fontweight='bold')
plt.tight_layout()
plt.savefig('model_comparison.png', dpi=150)
plt.show()



print("Results & Evaluation Complete")




LAST_CONV_LAYER = 'conv4_last'

def make_gradcam_heatmap(img_array, model, last_conv_layer_name):
    grad_model = tf.keras.models.Model(
        inputs=model.inputs,
        outputs=[model.get_layer(last_conv_layer_name).output,
                 model.output]
    )
    with tf.GradientTape() as tape:
        conv_outputs, predictions = grad_model(img_array)
        pred_index    = tf.argmax(predictions[0])
        class_channel = predictions[:, pred_index]

    grads        = tape.gradient(class_channel, conv_outputs)
    pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))
    heatmap      = conv_outputs[0] @ pooled_grads[..., tf.newaxis]
    heatmap      = tf.squeeze(heatmap)
    heatmap      = tf.maximum(heatmap, 0) / (tf.math.reduce_max(heatmap) + 1e-8)
    return heatmap.numpy(), predictions.numpy()


def display_gradcam(img_path, model, last_conv_layer_name, class_names):
    img           = load_img(img_path, target_size=(128, 128))
    img_array     = img_to_array(img)
    img_exp       = np.expand_dims(img_array / 255.0, axis=0)

    heatmap, preds  = make_gradcam_heatmap(img_exp, model, last_conv_layer_name)
    heatmap_colored = cm.jet(np.uint8(255 * heatmap))[:, :, :3]
    heatmap_colored = tf.image.resize(heatmap_colored, (128, 128)).numpy()
    superimposed    = np.clip(heatmap_colored * 0.4 + img_array / 255.0, 0, 1)

    predicted_cls = class_names[np.argmax(preds)]
    confidence    = np.max(preds) * 100

    fig, axes = plt.subplots(1, 2, figsize=(10, 4))
    axes[0].imshow(img_array / 255.0)
    axes[0].set_title('Original MRI')
    axes[0].axis('off')
    axes[1].imshow(superimposed)
    axes[1].set_title(f'Grad-CAM | {predicted_cls} ({confidence:.1f}%)')
    axes[1].axis('off')
    plt.tight_layout()
    plt.savefig(f'gradcam_{predicted_cls}.png', dpi=150)
    plt.show()


for cls in classes:
    sample_img_path = os.path.join(test_dir, cls,
                        os.listdir(os.path.join(test_dir, cls))[0])
    print(f"\nGrad-CAM — {cls}")
    display_gradcam(sample_img_path, cnn_model, LAST_CONV_LAYER, class_names)

print("Grad-CAM Complete")



cnn_f1    = f1_score(test_gen.classes, cnn_preds,    average='macro')
svm_f1    = f1_score(y_test_enc,       svm_preds,    average='macro')

print("\n" + "=" * 60)
print("        FINAL MODEL PERFORMANCE SUMMARY")
print("=" * 60)
print(f"{'Model':<25} {'Accuracy':>10} {'F1 (Macro)':>12}")
print("-" * 60)
print(f"{'SVM (Baseline)':<25} {svm_acc:>10.2%} {svm_f1:>12.4f}")
print(f"{'Custom CNN':<25} {cnn_acc:>10.2%} {cnn_f1:>12.4f}")
print("=" * 60)
print("\n Project Complete!")
```
